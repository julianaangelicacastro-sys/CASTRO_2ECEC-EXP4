# A. VISAYAS COMMUNICATION DATAFRAME

```python
# students whose Hometown is Visayas and whose Track is Communication

VisComm = df.loc[(df["Hometown"] == "Visayas")&(df["Track"] == "Communication")].copy()
print("Students whose Hometown is Visayas and whose Track is Communication")
VisComm
```

```python
# Retain only Name, Gender, Math, Electronics and Average

VisComm["Average"]=VisComm[["Math","Electronics","GEAS","Communication"]].mean(axis=1)
VisComm = VisComm.filter(items=['Name','Gender','Math','Electronics','Average'])
print("Students whose Hometown is Visayas and whose Track is Communication")
VisComm
```

```python
# Number of rows

len(VisComm)
```

# B. VISAYAS FEMALE DATAFRAME

```python
# Students whose Hometown is Visayas and whose Gender is Female

VisFemale = df.loc[(df["Hometown"] == "Visayas")&(df["Gender"] == "Female")].copy()
print("Students whose Hometown is Visayas and whose Gender is Female")
VisFemale
```

```python
# Retain only Name, Track, GEAS, Electronics and Average

VisFemale["Average"]=VisFemale[["Math","Electronics","GEAS","Communication"]].mean(axis=1)
VisFemale = VisFemale.filter(items=['Name','Track','GEAS','Electronics','Average'])
print("Students whose Hometown is Visayas and whose Gender is Female")
VisFemale
```

```python
# whose Average is at least 60

print("Students whose Average is at least 60")
VisFemale.loc[(VisFemale['Average']>=60)]
```

# C. CATEGORY-AVERAGE VISUALIZATION

```python
# mean of average for every category

df["Average"]=df[["Math","Electronics","GEAS","Communication"]].mean(axis=1)
print("mean of average for every category")

df
```

```python
# mean of Average for Track

track_average=df.groupby("Track")["Average"].mean().reset_index()
print("mean of Average for Track")
track_average
```

```python
# mean of Average for Gender

gender_average=df.groupby("Gender")["Average"].mean().reset_index()
print("mean of Average for Gender")
gender_average
```

```python
# mean of Average for Hometown

hometown_average=df.groupby("Hometown")["Average"].mean().reset_index()
print("mean of Average for Hometown")
hometown_average
```

```python
import matplotlib.pyplot as plt
```

```python
# mean Average by Track, by Gender, and by Hometown

fig,axes= plt.subplots(1,3,figsize=(12,5),sharey=True)

track_average.plot.bar(x="Track", y="Average", ax=axes[0], color="red")
axes[0].set_title("mean Average by Track")
axes[0].set_xlabel("Track")
axes[0].set_ylabel("mean Average")

gender_average.plot.bar(x="Gender", y="Average", ax=axes[1], color="Blue")
axes[1].set_title("mean Average by Gender")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("mean Average")

hometown_average.plot.bar(x="Hometown", y="Average", ax=axes[2], color="Green")
axes[2].set_title("mean Average by Hometown")
axes[2].set_xlabel("Hometown")
axes[2].set_ylabel("mean Average")
```
